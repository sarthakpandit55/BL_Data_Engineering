```python
import numpy as np
```

```python
np.random.seed(1)
vitals = np.random.randint(60, 100, size=(60, 24))
print(vitals)
```

```python
average_heart_rate = np.mean(vitals, axis=1)
print(f"Hourly average heart rate:- \n{average_heart_rate.round()}")
```

```python
abnormal_value = vitals[(vitals < 60) | (vitals > 100)]
print(f"Detected Abnormal reading out of range are :- {abnormal_value}")
```

```python
vitals = vitals.astype(float)
vitals[5,2] = np.nan
vitals[0,1] = np.nan
for i in range(1, vitals.shape[1]):
    mask = np.isnan(vitals[:,i])
    print(mask)

```
