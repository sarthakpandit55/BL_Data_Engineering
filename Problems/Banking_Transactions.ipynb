```python
import numpy as np
```

```python
np.random.seed(1)
data = np.random.randint(-1000,1000, size=(50, 20))
print(data)
```

```python
credit_transaction = data[data>0]
print(f"Credited transaction :- \n{credit_transaction}\n\n")
debit_transaction = data[data<0]
print(f"Debited transaction :- \n{debit_transaction}")
```

```python
daily_net_balance = np.sum(data,axis=0)
print(f"Daily net balance :- \n{daily_net_balance}")
```

```python
negative_account = (np.diff(data, axis=1) < 0).all(axis=1)
print(f"Negative account :- \n{negative_account}")
```
