```python
import numpy as np
```

```python
# Set 3 (Weather Monitoring System)
# NumPy (Intermediate)
#  A weather station stores hourly temperature readings for 7 days in a NumPy array of shape (7, 24).
# Task:
# Find the daily maximum and minimum temperature
#
#
# Identify the day with the largest temperature variation
#
# Replace outliers beyond ±2 standard deviations with the mean temperature
```

```python
np.random.seed(1)
data = np.random.randint(-2, 42, size=(7,24))
print(data)
```

```python
min_temp = data.min(axis=1)
print(f"Minimum temperature of each day:- {min_temp}")
max_temp = data.max(axis=1)
print(f"Maximum temperature of each day:- {max_temp}")
```

```python
temp_var = max_temp - min_temp
print(f"Variation in temperature of each day:- \n{temp_var}")
print(f"Day '{temp_var.argmax() + 1}' has the largest temperature variation of the day")
```

```python
mean_temp = data.mean()
std_temp = data.std()

lo = mean_temp - 2*std_temp
hi = mean_temp + 2*std_temp

outliner_mask = (data < lo) | (data > hi)
data_cleaned = np.where(outliner_mask, mean_temp, data)
print(data_cleaned)
```
