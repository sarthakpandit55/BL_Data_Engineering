```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from faker import Faker
```

```python
fake = Faker()
Faker.seed(16)

data = {
    "date" : [fake.date_between(start_date='-6m', end_date='today') for _ in range(5)],
    "city" : [fake.city() for _ in range(5)],
    "temperature" : [fake.random_int(min = -1, max = 45) for _ in range(5)],
    "humidity" : [fake.random_int(min = 30, max = 50) for _ in range(5)],
}

df = pd.DataFrame(data)
df
```
