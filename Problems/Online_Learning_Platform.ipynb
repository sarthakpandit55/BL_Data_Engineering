```python
import numpy as np
```

```python
np.random.seed(1)
course_data = np.random.randint(0, 61, size=(14, 6))
print(course_data)
```

```python
time_per_course = course_data.sum(axis=0)
print(time_per_course)
```

```python
threshold = time_per_course.mean()
print(threshold)
```

```python
engagement = np.argsort(time_per_course)[::-1]
print(engagement)
for rank, idx in enumerate(engagement):
    print(f"Time engaged on course {idx+1} is {time_per_course[idx]}")
```
