```python
import numpy as np
```

```python
np.random.seed(45)
```

```python
rating = np.random.randint(1,6, size=(100,4))
rating
```

```python
maximum = rating.max()
minimum = rating.min()
normalize = (rating-minimum) / (maximum-minimum)
print(f"Normalized ratting of employee :-\n{normalize}")
```

```python
average_rating = rating.mean(axis=1)
print(f"Employee average rating is:- \n {average_rating}")
```

```python
company_mean = average_rating.mean()
print(f"Company average rating is:- \n {company_mean}")

above_company_mean = np.where(average_rating>company_mean)
print(f"Employee with greater mean than company mean:- \n{above_company_mean}")
```
